This program takes a sentence stored in a variable and performs simple text analysis. It prints the original sentence, counts how many characters it has, finds all unique characters, splits the sentence into individual words, and then counts and displays those words. This helps beginners see how Python can analyze and process text step by step.



```python

# Sample text stored in a variable called "text"
# A variable is like a labeled box that holds a value.
text = "Natural Language Processing (NLP) enables computers to understand human language."

# Print a label so the output is easier to read
print("Original Text:")

# Print the actual text stored in the variable
print(text)

# Print a blank line, then the length of the text
# \n means "new line" (move to the next line)
# len(text) counts how many characters are in the string (including spaces and punctuation)
print("\nLength of the text:", len(text))

# Find all unique characters in the text
# set(text) creates a set, which is a collection that automatically removes duplicates
unique_characters = set(text)

# Print a blank line, then show the unique characters
print("\nUnique characters:", unique_characters)

# Count how many words are in the text
# text.split() splits the string into a list of words using spaces as separators
words = text.split()

# Print a blank line, then the number of words
# len(words) counts how many items (words) are in the list
print("\nNumber of words:", len(words))

# Print a blank line, then a label for the word list
print("\nWords in the text:")

# Print the list of words so we can see each word separately
print(words)
```

This program cleans and prepares a sentence for basic text processing. It first converts all letters to lowercase, then removes punctuation, and finally splits the text into individual words (tokens). The code prints each step so you can see the lowercase text, the text without punctuation, and the list of word tokens.

```python
# We import the built-in 'string' module.
# This module has helpful constants like string.punctuation (a list of common punctuation marks).
import string

# Sample text stored in a variable called "text"
# A variable is like a box with a label that stores some information.
text = "Natural Language Processing (NLP) enables computers to understand human language."

# Convert all letters in the text to lowercase.
# This helps make the text more consistent (for example, "Natural" and "natural" become the same).
text = text.lower()

# Print a label so we know what we are looking at
print("Lowercased Text:")

# Print the lowercase version of the text
print(text)

# Remove punctuation from the text.
# string.punctuation is a string containing characters like . , ! ? etc.
# str.maketrans('', '', string.punctuation) creates a "translation table"
# that tells Python to remove all punctuation characters from the text.
text = text.translate(str.maketrans('', '', string.punctuation))

# Print a blank line and a label for the text without punctuation
print("\nText without Punctuation:")

# Print the text after removing punctuation marks
print(text)

# Tokenize (split) the text into separate words.
# text.split() breaks the string into a list of words using spaces as separators.
tokens = text.split()

# Print a blank line and a label for the tokens
print("\nTokens:")

# Print the list of tokens (each word is one item in the list)
print(tokens)
```

This code takes a sentence, splits it into individual words, and then removes common English stop words like “the,” “is,” and “and.” It uses NLTK’s built-in stopword list to filter out these words, leaving only the meaningful terms. Finally, it prints both the original tokens and the cleaned-up list without stop words.

```python
# Import the NLTK library, which is used for working with human language data (text)
import nltk

# Import the list of common "stop words" (like "the", "is", "and") from NLTK
from nltk.corpus import stopwords

# Download the stopwords data the first time you run this code
# This tells NLTK to get the list of stopwords from the internet and save it locally
nltk.download('stopwords')

# Sample text stored in a variable called "text"
# A variable is like a labeled box that stores information
text = "Natural Language Processing enables computers to understand human language."

# Tokenize (split) the text into separate words
# text.split() breaks the sentence into a list of words using spaces
tokens = text.split()

# Get the set of English stop words
# Using a set makes it faster to check if a word is in the stopword list
stop_words = set(stopwords.words('english'))

# Create a new list called filtered_tokens that:
# - keeps only the words that are NOT stop words
# - word.lower() makes the word lowercase before checking (so "The" and "the" are treated the same)
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

# Print the original list of words (tokens) before removing stop words
print("Original Tokens:")
print(tokens)

# Print a blank line (\n) and then the filtered list of words
# These are the words with common stop words removed
print("\nFiltered Tokens:")
print(filtered_tokens)
```

This code takes a sentence, splits it into individual words, and uses the Porter Stemmer to reduce each word to its root form (called a “stem”). It then prints both the original words and their stemmed versions, showing how stemming simplifies words for text processing.

```python
# Import the PorterStemmer class from NLTK.
# A "stemmer" reduces words to their basic form (called the "stem").
from nltk.stem import PorterStemmer

# Sample text stored in a variable called "text"
# A variable is like a labeled box that stores some information.
text = "Natural Language Processing enables computers to understand human language."

# Tokenize (split) the text into separate words.
# text.split() breaks the sentence into a list of words using spaces.
tokens = text.split()

# Create (initialize) a stemmer object that we can use to stem words.
stemmer = PorterStemmer()

# Go through each word in tokens and apply the stemmer to it.
# stemmer.stem(word) returns the stem (base form) of the word.
# We store all the stemmed words in a new list called stemmed_tokens.
stemmed_tokens = [stemmer.stem(word) for word in tokens]

# Print the original list of words (before stemming)
print("Original Tokens:")
print(tokens)

# Print a blank line (\n) and then the list of stemmed words
print("\nStemmed Tokens:")
print(stemmed_tokens)
```

This code takes a sentence, splits it into individual words, and uses NLTK’s WordNet lemmatizer to convert each word into its dictionary form (lemma). After downloading the WordNet data, it processes each word and prints both the original tokens and their lemmatized versions, showing how lemmatization makes words more standardized for language processing.

```python
# Import the WordNetLemmatizer class from NLTK.
# A "lemmatizer" changes a word to its dictionary form (lemma),
# for example: "running" -> "run", "better" -> "good" (sometimes).
from nltk.stem import WordNetLemmatizer

# Import the main NLTK library so we can download needed data
import nltk

# Download the WordNet data the first time you run this code.
# WordNet is a large dictionary that helps the lemmatizer know word forms.
nltk.download('wordnet')

# Download additional language data needed by WordNet
nltk.download('omw-1.4')

# Sample text stored in a variable called "text"
# A variable is like a labeled box that stores information.
text = "Natural Language Processing enables computers to understand human language."

# Tokenize (split) the text into separate words.
# text.split() breaks the sentence into a list of words using spaces.
tokens = text.split()

# Create (initialize) a lemmatizer object that we can use on words.
lemmatizer = WordNetLemmatizer()

# Go through each word in tokens and apply the lemmatizer to it.
# lemmatizer.lemmatize(word) returns the lemma (dictionary form) of the word.
# We store all the lemmatized words in a new list called lemmatized_tokens.
lemmatized_tokens = [lemmatizer.lemmatize(word) for word in tokens]

# Print the original list of words (before lemmatization)
print("Original Tokens:")
print(tokens)

# Print a blank line (\n) and then the list of lemmatized words
print("\nLemmatized Tokens:")
print(lemmatized_tokens)
```

This code cleans and simplifies a sentence step by step for NLP. It first lowercases the text and removes punctuation, then splits it into words. Next, it removes common English stop words (like “the” and “is”). Finally, it applies stemming and lemmatization to the remaining words to reduce them to their base or dictionary forms, and prints the cleaned tokens at each stage.


```python
# Import the list of common stop words (like "the", "is", "and") from NLTK
from nltk.corpus import stopwords

# Import two tools from NLTK:
# - PorterStemmer: reduces words to their basic stem (e.g., "running" -> "run")
# - WordNetLemmatizer: changes words to their dictionary form (lemma)
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Import the main NLTK library so we can download needed data
import nltk

# Download the list of stop words (only needed the first time you run this code)
nltk.download('stopwords')

# Download the WordNet data used by the lemmatizer (only needed the first time)
nltk.download('wordnet')

# Sample text stored in a variable called "text"
# A variable is like a labeled box that stores information.
text = "Natural Language Processing enables computers to understand human language."

# Convert all letters in the text to lowercase.
# This makes the text consistent, so "Natural" and "natural" are treated the same.
text = text.lower()

# Import the string module so we can access a list of punctuation characters
import string

# Remove punctuation from the text.
# string.punctuation is a string of characters like . , ! ? etc.
# str.maketrans('', '', string.punctuation) creates a translation table
# that tells Python to delete all punctuation characters from the text.
text = text.translate(str.maketrans('', '', string.punctuation))

# Tokenize (split) the text into separate words.
# text.split() breaks the sentence into a list of words using spaces.
tokens = text.split()

# Get the set of English stop words.
# A set is used because it allows fast checking of "is this word in the list?"
stop_words = set(stopwords.words('english'))

# Remove stop words from our list of tokens.
# This list comprehension means:
# "for each word in tokens, keep it only if it is NOT in stop_words".
filtered_tokens = [word for word in tokens if word not in stop_words]

# Create (initialize) the stemmer and lemmatizer objects that we will use.
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Apply stemming and then lemmatization to each filtered word.
# For each word in filtered_tokens:
#   1. stemmer.stem(word) gets the stem (basic form)
#   2. lemmatizer.lemmatize(...) changes that stem to a dictionary form if possible
# The final results are stored in a new list called processed_tokens.
processed_tokens = [lemmatizer.lemmatize(stemmer.stem(word)) for word in filtered_tokens]

# Print the cleaned version of the text (lowercased and without punctuation)
print("Original Text:")
print(text)

# Print a blank line and then the tokens after removing stop words
print("\nFiltered Tokens (Stop Words Removed):")
print(filtered_tokens)

# Print another blank line and then the tokens after stemming and lemmatization
print("\nProcessed Tokens (Stemmed and Lemmatized):")
print(processed_tokens)
```

This code uses Python’s regular expression module (re) to search a sentence for the word “fox.” It defines a simple pattern, checks whether the pattern appears in the text using re.search(), and then prints either the matched word or a message saying no match was found.

```python
# Import the 're' module.
# This module lets us work with regular expressions (patterns used to search text).
import re

# Sample text stored in a variable called "text"
# A variable is like a labeled box that holds information.
text = "The quick brown fox jumps over the lazy dog."

# Define a pattern to search for.
# r"fox" means we are looking for the exact word "fox" in the text.
# The 'r' before the string makes it a "raw string," which is useful for regex.
pattern = r"fox"

# Use re.search() to look for the pattern in the text.
# If the pattern is found, re.search() returns a match object.
# If it is not found, it returns None.
match = re.search(pattern, text)

# Check if a match was found
if match:
    # match.group() gives us the actual text that matched the pattern
    print("Match found:", match.group())
else:
    # This runs if re.search() did not find the pattern in the text
    print("No match found.")
```

This code scans a sentence for email addresses using a regular expression pattern. It uses re.findall() to extract every substring that looks like an email (such as support@example.com), then prints the list of all matched email addresses found in the text.

```python
# Import the 're' module.
# This module lets us use regular expressions (patterns) to search inside text.
import re

# Sample text that contains two email addresses
# We store the text inside a variable called "text".
text = "Please contact us at support@example.com or sales@example.com for further information."

# Define a regular expression (regex) pattern to match email addresses.
# r"..." means this is a raw string, which is helpful when writing regex.
# \b                = word boundary (start or end of a word)
# [A-Za-z0-9._%+-]+ = one or more allowed characters before the @ symbol
# @                 = the @ symbol in an email
# [A-Za-z0-9.-]+    = one or more allowed characters for the email domain
# \.                = a dot (.)
# [A-Za-z]{2,}      = at least two letters for the domain ending (like com, org)
# \b                = word boundary
pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b"

# Use re.findall() to find all pieces of text that match the pattern.
# It returns a list of all matching email addresses found in the text.
emails = re.findall(pattern, text)

# Print a label so we know what we are looking at
print("Extracted Email Addresses:")

# Print the list of email addresses that were found
print(emails)
```

This code uses a regular expression to find phone numbers in the format (123) 456-7890 inside a sentence. It searches the text with re.findall(), collects all matching phone numbers, and prints them as a list.

```python
# Import the 're' module.
# This module lets us use regular expressions (patterns) to search inside text.
import re

# Sample text that contains two phone numbers.
# We store the text in a variable called "text".
text = "Contact us at (123) 456-7890 or (987) 654-3210."

# Define a regular expression (regex) pattern to match phone numbers
# in the format: (123) 456-7890
#
# r"..."   = raw string (so \ is treated correctly in the pattern)
# \(       = a literal "(" character
# \d{3}    = exactly 3 digits
# \)       = a literal ")" character
# (space)  = a space character
# \d{3}    = exactly 3 digits
# -        = a literal "-" character
# \d{4}    = exactly 4 digits
pattern = r"\(\d{3}\) \d{3}-\d{4}"

# Use re.findall() to search the text for ALL matches of the pattern.
# It returns a list of all phone numbers that match the pattern.
phone_numbers = re.findall(pattern, text)

# Print a label so we know what is being shown
print("Extracted Phone Numbers:")

# Print the list of phone numbers that were found in the text
print(phone_numbers)
```

This code searches a sentence for the word “fox” and replaces every occurrence with “cat” using re.sub(). It then prints the updated sentence, showing how regular expressions can be used for simple text replacement.

```python
# Import the 're' module.
# This module lets us work with regular expressions (patterns) to search and replace text.
import re

# Sample text stored in a variable called "text"
# A variable is like a labeled box that holds some information.
text = "The quick brown fox jumps over the lazy dog. The fox is clever."

# Define a pattern (regular expression) to match the word "fox"
# Here, the pattern is just the exact word "fox".
pattern = r"fox"

# Use re.sub() to replace all occurrences of "fox" with "cat" in the text.
# re.sub(pattern, "cat", text):
#   - pattern: what we are looking for ("fox")
#   - "cat": what we want to replace it with
#   - text: the original text
new_text = re.sub(pattern, "cat", text)

# Print a label so we know what is being shown
print("Modified Text:")

# Print the text after replacing "fox" with "cat"
print(new_text)
```

This code uses a regular expression to find dates in a sentence. It looks for two formats—YYYY-MM-DD (like 2022-08-15) and DD/MM/YYYY (like 15/08/2022)—using re.findall(), then prints a list of all the dates it finds in the text.

```python
# Import the 're' module.
# This module lets us use regular expressions (patterns) to search inside text.
import re

# Sample text with two dates in different formats.
# We store the text in a variable called "text".
text = "The event is scheduled for 2022-08-15. Another event is on 15/08/2022."

# Define a regular expression (regex) pattern to match dates.
#
# r"..."           = raw string (so backslashes \ are handled correctly)
# \d{4}-\d{2}-\d{2}  matches dates like 2022-08-15   (YYYY-MM-DD)
#   \d    = a digit (0–9)
#   {4}   = exactly 4 digits
#   -     = a literal dash
#   {2}   = exactly 2 digits (for month and day)
#
# |                 = OR (match either the left side or the right side)
#
# \d{2}/\d{2}/\d{4}  matches dates like 15/08/2022   (DD/MM/YYYY)
#   /    = a literal slash
pattern = r"\d{4}-\d{2}-\d{2}|\d{2}/\d{2}/\d{4}"

# Use re.findall() to search the text for ALL matches of the pattern.
# It returns a list of all substrings that look like dates.
dates = re.findall(pattern, text)

# Print a label so we know what is being shown
print("Extracted Dates:")

# Print the list of dates that were found in the text
print(dates)
```

This code uses a regular expression to scan a sentence for hashtags. It finds any word that begins with # (like #excited or #tech) using re.findall() and prints all the hashtags it detects in the text.

```python
# Import the 're' module.
# This module lets us use regular expressions (patterns) to search inside text.
import re

# Sample text that contains some hashtags (like on social media)
# We store the text in a variable called "text".
text = "Loving the new features of this product! #excited #newrelease #tech"

# Define a regular expression (regex) pattern to match hashtags.
#
# r"..."  = raw string (so backslashes \ are treated correctly)
# #       = the hashtag symbol (we want to match it exactly)
# \w+     = one or more "word" characters (letters, numbers, or underscore)
#          So this matches things like #excited, #newrelease, #tech
pattern = r"#\w+"

# Use re.findall() to search the text for ALL parts that match the pattern.
# It returns a list of all hashtags found in the text.
hashtags = re.findall(pattern, text)

# Print a label so we know what is being shown
print("Extracted Hashtags:")

# Print the list of hashtags that were found
print(hashtags)
```

This code uses NLTK’s word_tokenize() function to split a sentence into individual tokens, such as words and punctuation marks. After downloading the required tokenizer data, it processes the sample text and prints the list of word tokens.

```python
# Import the NLTK library, which is used for working with human language data (text)
import nltk

# Download the tokenizer resources that NLTK needs.
# This only needs to be done once on a new computer.
nltk.download('punkt')        # Main tokenizer data
nltk.download('punkt_tab')    # Extra data needed for some environments

# Import the word_tokenize function.
# This function splits a sentence into individual words and punctuation marks.
from nltk.tokenize import word_tokenize

# Sample text stored in a variable called "text"
# A variable is like a labeled box that holds information.
text = "Natural Language Processing enables computers to understand human language."

# Use word_tokenize() to break the sentence into separate tokens.
# A "token" is usually a word or punctuation mark.
tokens = word_tokenize(text)

# Print a label so we know what we are looking at
print("Word Tokens:")

# Print the list of tokens generated by NLTK
print(tokens)
```

This code uses SpaCy’s English language model to break a sentence into individual tokens. After loading the model and processing the text, it extracts each token (word or punctuation mark) from the SpaCy doc object and prints the full list of tokens.

```python
# Import the SpaCy library.
# SpaCy is a powerful library for Natural Language Processing (NLP),
# which means working with and analyzing human language.
import spacy

# Load the small English language model.
# This model knows basic rules about English, like how to split text into words.
# "en_core_web_sm" must be installed first (usually by running:
# python -m spacy download en_core_web_sm in your terminal).
nlp = spacy.load("en_core_web_sm")

# Sample text stored in a variable called "text"
# A variable is like a labeled box that holds information.
text = "Natural Language Processing enables computers to understand human language."

# Use the SpaCy model to process the text.
# This returns a "doc" object that contains tokens and other language features.
doc = nlp(text)

# Create a list of tokens (words and punctuation) from the doc.
# token.text gives the original text for each token.
tokens = [token.text for token in doc]

# Print a label so we know what we are looking at
print("Word Tokens:")

# Print the list of tokens
print(tokens)
```

This code uses SpaCy’s English language model to split a paragraph into individual sentences. After processing the text with SpaCy, it extracts each sentence from the doc object and prints them as a list.

```python
# Import the SpaCy library.
# SpaCy is a powerful library for working with and analyzing human language (NLP).
import spacy

# Load the small English language model.
# This model knows how to split text into words and sentences, and more.
nlp = spacy.load("en_core_web_sm")

# Sample text stored in a variable called "text"
# A variable is like a labeled box that stores information.
text = "Natural Language Processing enables computers to understand human language. It is a fascinating field."

# Use the SpaCy model to process the text.
# The result is a "doc" object that contains sentences, words, and other info.
doc = nlp(text)

# Extract each sentence from the doc.
# doc.sents is a sequence of sentence objects.
# sent.text gives the original text of each sentence.
sentences = [sent.text for sent in doc.sents]

# Print a label so we know what we are looking at
print("Sentences:")

# Print the list of sentences found in the text
print(sentences)
```
