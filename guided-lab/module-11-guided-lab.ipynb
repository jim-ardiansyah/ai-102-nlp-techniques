This program creates a very basic rule-based chatbot that responds using a small set of predefined answers. It waits for the user to type a message, checks if the message matches one of the known phrases (like “hello” or “bye”), and replies with the corresponding response. If the message is not recognized, the bot returns a default “I don’t understand” message. The chatbot continues the conversation in a loop until the user types "exit", which ends the program.

```python
# ----------------------------------------------------------
# SIMPLE RULE-BASED CHATBOT
# ----------------------------------------------------------
# This code creates a very basic chatbot that:
# 1) Waits for the user to type a message.
# 2) Checks if the message matches one of the known phrases.
# 3) Responds with a pre-written answer.
# 4) Keeps chatting until the user types "exit".
# ----------------------------------------------------------

def chatbot_response(user_input):
    """
    Return a response based on the user's input.
    If the input is not recognized, return a default message.
    """

    # A dictionary that maps user messages (keys) to chatbot replies (values).
    # If the user types one of these keys, the bot will answer with the matching value.
    responses = {
        "hello": "Hello! How can I assist you today?",
        "hi": "Hi there! What can I do for you?",
        "how are you?": "I'm just a chatbot, but I'm here to help you!",
        "what is your name?": "I am ChatBot, your virtual assistant.",
        "bye": "Goodbye! Have a great day!"
    }

    # Convert the user input to lowercase so that:
    # "Hello", "HELLO", and "hello" are treated the same.
    user_input = user_input.lower()

    # Look up the user input in the responses dictionary.
    # responses.get(key, default_value) returns:
    #   - responses[key] if key exists
    #   - default_value if key does not exist
    return responses.get(
        user_input,
        "I'm sorry, I don't understand that. Can you please rephrase?"
    )

# ----------------------------------------------------------
# MAIN CHAT LOOP
# ----------------------------------------------------------
# This loop keeps asking the user for input until they type "exit".
while True:
    # Ask the user to type something.
    user_input = input("You: ")

    # If the user types "exit" (in any mix of upper or lower case), stop the loop.
    if user_input.lower() == "exit":
        print("ChatBot: Goodbye! Have a great day!")
        break  # break exits the while loop

    # Otherwise, get a response from the chatbot.
    response = chatbot_response(user_input)

    # Print the chatbot's response.
    print(f"ChatBot: {response}")
```

This program creates a hybrid customer service chatbot that can answer questions in two ways. First, it checks if the user’s question matches a small list of predefined FAQs (such as store hours, location, or return policy). If it finds a match, it replies with a fixed, prewritten answer. If the question is something else, the chatbot uses the OpenAI GPT model to generate a more flexible, natural response. The program runs in a loop, allowing the user to chat with the bot until they type “exit”, which ends the conversation.

```python
# ----------------------------------------------------------
# SIMPLE CUSTOMER SERVICE CHATBOT (RULES + OPENAI GPT)
# ----------------------------------------------------------
# This chatbot answers customer questions in two ways:
# 1) If the question matches a known FAQ (store hours, location, etc.),
#    it returns a fixed, pre-written answer.
# 2) If the question is something else, it calls the OpenAI API (GPT)
#    to generate a more flexible, natural response.
# ----------------------------------------------------------

import openai  # Library used to talk to the OpenAI API

# ----------------------------------------------------------
# 1. SET YOUR OPENAI API KEY
# ----------------------------------------------------------
# This key tells OpenAI who you are and lets you use their models.
# IMPORTANT:
# - Never share your real API key in public code or GitHub.
# - Replace 'your-api-key' with your actual key when testing privately.
openai.api_key = 'your-api-key'

# ----------------------------------------------------------
# 2. DEFINE THE CHATBOT FUNCTION
# ----------------------------------------------------------
def customer_service_chatbot(user_input):
    """
    Returns a reply to the user's question.
    First, it checks if the question is in a list of common FAQs.
    If not, it asks the OpenAI GPT model to generate a response.
    """

    # Predefined (hard-coded) responses for common customer questions.
    # The keys are questions (in lowercase).
    # The values are the answers.
    predefined_responses = {
        "what are your store hours?": "Our store is open from 9 AM to 9 PM, Monday to Saturday.",
        "where is your store located?": "Our store is located at 123 Main Street, Anytown, USA.",
        "what is your return policy?": "You can return any item within 30 days of purchase with a receipt."
    }

    # Convert the user input to lowercase so:
    # "What are your store hours?" and "what are your store hours?"
    # are treated the same.
    user_input_lower = user_input.lower()

    # ------------------------------------------------------
    # 2.1 CHECK IF QUESTION MATCHES A PREDEFINED ANSWER
    # ------------------------------------------------------
    if user_input_lower in predefined_responses:
        # If the question is in our FAQ dictionary, return that answer.
        return predefined_responses[user_input_lower]

    # ------------------------------------------------------
    # 2.2 FALL BACK TO GPT FOR OTHER QUESTIONS
    # ------------------------------------------------------
    # If the question is NOT in our predefined list,
    # we ask the OpenAI model to generate a response.
    #
    # openai.Completion.create(...) sends a request to the model.
    # Parameters:
    #   engine      -> which model to use (e.g., "gpt-4")
    #   prompt      -> the text we send to the model as input
    #   max_tokens  -> maximum length of the response
    #   n           -> number of responses to generate (we just want 1)
    #   stop        -> where the model should stop (None = no special stop sequence)
    #   temperature -> how "creative" the response is (0 = strict, 1 = very random)
    response = openai.Completion.create(
        engine="gpt-4",
        prompt=f"User: {user_input}\nCustomer Service Bot:",
        max_tokens=150,
        n=1,
        stop=None,
        temperature=0.7,
    )

    # The response object may contain multiple choices.
    # We take the first one (choices[0]) and get its text.
    # .strip() removes extra spaces/newlines at the start and end.
    return response.choices[0].text.strip()

# ----------------------------------------------------------
# 3. MAIN CHAT LOOP (INTERACTIVE MODE)
# ----------------------------------------------------------
# This loop allows the user to chat with the bot from the terminal.
while True:
    # Ask the user to type a question.
    user_input = input("You: ")

    # If the user types "exit" (any case), end the chat.
    if user_input.lower() == "exit":
        print("Customer Service Bot: Thank you for reaching out! Have a great day!")
        break  # Exit the while loop and end the program

    # Otherwise, get a reply from the chatbot.
    response = customer_service_chatbot(user_input)

    # Print the chatbot's reply.
    print(f"Customer Service Bot: {response}")
```

This program builds a simple retrieval-based chatbot that chooses the best reply from a small list of predefined responses. It first converts all example responses (the corpus) into TF-IDF vectors, then converts the user’s input into another TF-IDF vector. Using cosine similarity, it compares the user input to every response and selects the one that is most similar. The chat runs in a loop in the terminal until the user types "exit", at which point the program ends.

```python
# ----------------------------------------------------------
# RETRIEVAL-BASED CHATBOT WITH TF-IDF + COSINE SIMILARITY
# ----------------------------------------------------------
# This chatbot does NOT generate new sentences.
# Instead, it:
# 1) Stores a small list of example responses (a "corpus").
# 2) Turns each response into a TF-IDF vector (numbers).
# 3) Turns the user's input into a TF-IDF vector.
# 4) Finds which stored response is most similar to the user's input.
# 5) Returns that response.
# ----------------------------------------------------------

# TfidfVectorizer:
#   - converts text into numerical vectors using TF-IDF
#   - TF-IDF = Term Frequency – Inverse Document Frequency
#   - words that are important in a sentence get higher scores
from sklearn.feature_extraction.text import TfidfVectorizer

# cosine_similarity:
#   - measures how similar two vectors are
#   - 1.0 means exactly the same direction, 0 means not similar
from sklearn.metrics.pairwise import cosine_similarity

# ----------------------------------------------------------
# 1. DEFINE A SMALL CORPUS OF RESPONSES
# ----------------------------------------------------------
# This is our "knowledge base" of possible chatbot replies.
# The chatbot will pick one of these sentences as its answer.
corpus = [
    "Hello! How can I assist you today?",
    "Hi there! What can I do for you?",
    "I'm just a chatbot, but I'm here to help you!",
    "I am ChatBot, your virtual assistant.",
    "Goodbye! Have a great day!"
]

# ----------------------------------------------------------
# 2. CREATE TF-IDF VECTORS FOR THE CORPUS
# ----------------------------------------------------------
# Create the vectorizer object.
vectorizer = TfidfVectorizer()

# fit_transform():
#   - fit: learn the vocabulary from 'corpus'
#   - transform: convert each sentence into a TF-IDF vector
#
# X is a 2D matrix where:
#   - each row represents one sentence from 'corpus'
#   - each column represents one word from the vocabulary
X = vectorizer.fit_transform(corpus)

# ----------------------------------------------------------
# 3. DEFINE THE RETRIEVAL-BASED CHATBOT FUNCTION
# ----------------------------------------------------------
def retrieval_based_chatbot(user_input):
    """
    Given a user input string, find and return the most similar
    response from the predefined corpus using TF-IDF and cosine similarity.
    """

    # Convert the user's input into a TF-IDF vector using
    # the same vectorizer we fitted on the corpus.
    user_input_vector = vectorizer.transform([user_input])

    # Compute cosine similarity between:
    #   - user_input_vector (1 row)
    #   - X (all corpus sentence vectors)
    #
    # 'similarities' will be a 2D array with shape (1, number_of_corpus_sentences)
    similarities = cosine_similarity(user_input_vector, X)

    # Find the index of the most similar sentence.
    # argmax() returns the position of the highest value.
    response_index = similarities.argmax()

    # Return the sentence from the corpus with the highest similarity score.
    return corpus[response_index]

# ----------------------------------------------------------
# 4. MAIN CHAT LOOP
# ----------------------------------------------------------
# This loop allows you to talk to the chatbot in the terminal.
while True:
    # Ask the user to type a message.
    user_input = input("You: ")

    # If user types "exit" (any capitalization), stop the chat.
    if user_input.lower() == "exit":
        print("ChatBot: Goodbye! Have a great day!")
        break  # Exit the while loop.

    # Get the chatbot's reply using the retrieval-based function.
    response = retrieval_based_chatbot(user_input)

    # Print the chatbot's response.
    print(f"ChatBot: {response}")
```

This program builds a hybrid chatbot that replies to user messages in two ways. First, it checks if the input exactly matches simple phrases like "hello", "hi", or "bye" and responds with a fixed, rule-based answer. If the input doesn’t match any of these, the chatbot uses TF-IDF and cosine similarity to compare the user’s message with a small set of predefined responses and returns the most similar one. The conversation runs in a loop in the terminal until the user types "exit" to stop the chat.

```python
# ----------------------------------------------------------
# HYBRID CHATBOT: RULE-BASED + RETRIEVAL-BASED RESPONSES
# ----------------------------------------------------------
# This chatbot combines two simple ideas:
#
# 1) Rule-based:
#    - If the user types an exact known phrase like "hello" or "hi",
#      the bot returns a fixed, pre-written answer.
#
# 2) Retrieval-based (TF-IDF + cosine similarity):
#    - If the input is not an exact match, the bot finds the most similar
#      sentence from a small list of responses (corpus) and returns it.
#
# This is called a *hybrid* chatbot because it mixes both approaches.
# ----------------------------------------------------------

# TfidfVectorizer converts text into numerical vectors based on word importance.
from sklearn.feature_extraction.text import TfidfVectorizer

# cosine_similarity measures how similar two vectors are (0 = no similarity, 1 = very similar).
from sklearn.metrics.pairwise import cosine_similarity

# ----------------------------------------------------------
# 1. DEFINE A SMALL CORPUS OF RESPONSES (FOR RETRIEVAL)
# ----------------------------------------------------------
# These are fallback responses the bot can choose from when the
# user input does not match the simple rule-based patterns.
corpus = [
    "I'm just a chatbot, but I'm here to help you!",
    "I am ChatBot, your virtual assistant.",
    "Goodbye! Have a great day!"
]

# Create a TF-IDF vectorizer object.
vectorizer = TfidfVectorizer()

# Learn the vocabulary from 'corpus' and convert each sentence into a TF-IDF vector.
# X is a matrix where each row corresponds to one sentence in 'corpus'.
X = vectorizer.fit_transform(corpus)

# ----------------------------------------------------------
# 2. DEFINE THE HYBRID CHATBOT FUNCTION
# ----------------------------------------------------------
def hybrid_chatbot(user_input):
    """
    Respond to the user using a hybrid strategy:
      1) If the input is a known simple phrase (like 'hello'), use a rule-based reply.
      2) Otherwise, use TF-IDF + cosine similarity to pick the most similar
         sentence from the corpus and return it.
    """

    # Rule-based responses for exact matches.
    # If the user types one of these keys (in lowercase), the bot returns the value.
    rule_based_responses = {
        "hello": "Hello! How can I assist you today?",
        "hi": "Hi there! What can I do for you?",
        "bye": "Goodbye! Have a great day!"
    }

    # Convert user input to lowercase so "Hello" and "hello" are treated the same.
    user_input_lower = user_input.lower()

    # ------------------------------------------------------
    # 2.1 CHECK RULE-BASED RESPONSES FIRST
    # ------------------------------------------------------
    if user_input_lower in rule_based_responses:
        # If the user input matches a key in rule_based_responses,
        # return that fixed response immediately.
        return rule_based_responses[user_input_lower]

    # ------------------------------------------------------
    # 2.2 FALL BACK TO RETRIEVAL-BASED RESPONSE
    # ------------------------------------------------------
    # If there's no exact rule-based match, we:
    #   1) Convert the user input into a TF-IDF vector.
    #   2) Compare it to all vectors in X (our corpus).
    #   3) Choose the most similar sentence from 'corpus'.

    # Convert the user input into a TF-IDF vector using the same vectorizer.
    user_input_vector = vectorizer.transform([user_input])

    # Compute cosine similarity between:
    #   - user_input_vector (1 row)
    #   - X (all corpus sentences)
    similarities = cosine_similarity(user_input_vector, X)

    # similarities is a 2D array of shape (1, len(corpus)).
    # similarities.argmax() gives the index of the highest similarity score.
    response_index = similarities.argmax()

    # Return the corpus sentence with the highest similarity.
    return corpus[response_index]

# ----------------------------------------------------------
# 3. MAIN CHAT LOOP
# ----------------------------------------------------------
# This loop allows you to talk to the chatbot in the terminal/console.
while True:
    # Ask the user to type a message.
    user_input = input("You: ")

    # If the user types "exit", we stop the conversation.
    if user_input.lower() == "exit":
        print("ChatBot: Goodbye! Have a great day!")
        break  # Exit the while loop, ending the program.

    # Otherwise, get a response from the hybrid_chatbot function.
    response = hybrid_chatbot(user_input)

    # Print the chatbot's reply.
    print(f"ChatBot: {response}")
```
