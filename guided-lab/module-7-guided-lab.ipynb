This code installs all the required Python libraries for implementing rule-based sentiment analysis and building machine learning and deep learning sentiment models. It includes tools like TextBlob and Afinn for rule-based sentiment analysis, TensorFlow and Keras for training neural networks, and Transformers and PyTorch for working with advanced pre-trained language models. These libraries together prepare the environment for multiple approaches to sentiment analysis.

```python
# ---------------------------------------------------------
# Import the TextBlob class from the textblob library.
# TextBlob helps us work with text and can analyze things like
# sentiment (how positive or negative a sentence is).
# ---------------------------------------------------------
from textblob import TextBlob

# ---------------------------------------------------------
# 1. CREATE SOME SAMPLE TEXT
# ---------------------------------------------------------
# This is the sentence we want to analyze.
# You can change this to any sentence you want.
text = "I love this product! It works wonderfully and the quality is excellent."

# ---------------------------------------------------------
# 2. CREATE A TEXTBLOB OBJECT
# ---------------------------------------------------------
# We pass our text into TextBlob.
# The result (called "blob") is an object that has useful
# functions and properties for analyzing the text.
blob = TextBlob(text)

# ---------------------------------------------------------
# 3. GET THE SENTIMENT
# ---------------------------------------------------------
# blob.sentiment returns two values:
#   - polarity: how positive or negative the text is
#               range is from -1.0 (very negative)
#               to 1.0 (very positive)
#   - subjectivity: how opinion based the text is
#                   range is from 0.0 (very objective, fact like)
#                   to 1.0 (very subjective, opinion based)
sentiment = blob.sentiment

# ---------------------------------------------------------
# 4. PRINT THE RESULTS
# ---------------------------------------------------------
print("Sentiment Analysis:")
# We use an f-string to nicely format the output.
print(f"Polarity: {sentiment.polarity}, Subjectivity: {sentiment.subjectivity}")
```

This code uses the Afinn library to perform simple, rule-based sentiment analysis on a sentence. It creates a sentiment analyzer, scores the text based on the emotional tone of its words, and then converts that numeric score into an easy-to-read label: Positive, Negative, or Neutral. This is an example of Creating Custom Rule-Based Sentiment Analyzers, where you define how numeric scores map to sentiment categories.

```python
from afinn import Afinn

# ---------------------------------------------------------
# 1. CREATE THE SENTIMENT ANALYZER
# ---------------------------------------------------------
# Here we make an Afinn object called "afinn".
# We will use this object to analyze the sentiment of text.
afinn = Afinn()

# ---------------------------------------------------------
# 2. WRITE SOME SAMPLE TEXT
# ---------------------------------------------------------
# This is the text we want to analyze.
# Feel free to change this sentence and see how the score changes.
text = "I hate the traffic in this city. It makes commuting a nightmare."

# ---------------------------------------------------------
# 3. GET THE SENTIMENT SCORE
# ---------------------------------------------------------
# afinn.score(text) looks at the text and returns a number:
#   - Positive number: the text is more positive.
#   - Negative number: the text is more negative.
#   - Zero: the text is neutral.
sentiment_score = afinn.score(text)

# ---------------------------------------------------------
# 4. TURN THE SCORE INTO A SENTIMENT LABEL
# ---------------------------------------------------------
# Now we translate the numeric score into a simple label
# that is easier to understand: "Positive", "Negative", or "Neutral".
if sentiment_score > 0:
    sentiment = "Positive"
elif sentiment_score < 0:
    sentiment = "Negative"
else:
    sentiment = "Neutral"

# ---------------------------------------------------------
# 5. PRINT THE RESULTS
# ---------------------------------------------------------
print("Sentiment Analysis:")
# Print the original text so we remember what we analyzed.
print(f"Text: {text}")
# Print the raw numeric score from Afinn.
print(f"Sentiment Score: {sentiment_score}")
# Print the easy-to-read sentiment label.
print(f"Sentiment: {sentiment}")
```

This code uses TF-IDF feature extraction to turn a small set of text reviews into numerical data that a machine learning model can understand. It creates a TF-IDF vectorizer, learns the important words in the corpus, and then transforms each sentence into a TF-IDF feature vector, showing how important each word is in each document. Finally, it prints the resulting TF-IDF matrix and the list of feature (word) names.

```python
# ---------------------------------------------------------
# Import the TfidfVectorizer class from scikit-learn.
# "TF-IDF" stands for "Term Frequency - Inverse Document Frequency".
# It is a way to turn text into numbers so that a computer can work with it.
#
# - Term Frequency (TF): How often a word appears in a document.
# - Inverse Document Frequency (IDF): How rare a word is across all documents.
#
# Words that appear a lot in one document, but not in all documents,
# get higher scores. Common words like "the" or "is" usually get low scores.
# ---------------------------------------------------------
from sklearn.feature_extraction.text import TfidfVectorizer

# ---------------------------------------------------------
# 1. CREATE A SMALL TEXT CORPUS
# ---------------------------------------------------------
# A "corpus" is just a collection (list) of documents.
# Here, each string in the list is like one review or one sentence.
corpus = [
    "I love this product! It's amazing.",
    "This is the worst service I have ever experienced.",
    "I am very happy with my purchase.",
    "I am disappointed with the quality of this item."
]

# ---------------------------------------------------------
# 2. INITIALIZE THE TF-IDF VECTORIZER
# ---------------------------------------------------------
# TfidfVectorizer will:
#   - break the text into words,
#   - count how often each word appears,
#   - calculate TF-IDF scores for each word in each document.
vectorizer = TfidfVectorizer()

# ---------------------------------------------------------
# 3. TRANSFORM THE TEXT DATA INTO TF-IDF FEATURES
# ---------------------------------------------------------
# fit_transform() does two things:
#   1) fit: learn the vocabulary (which words appear in the corpus)
#   2) transform: create the TF-IDF matrix using that vocabulary
#
# The result "X" is a matrix where:
#   - each row represents a document (a sentence from our corpus),
#   - each column represents a word from the vocabulary,
#   - each cell contains the TF-IDF score for that (document, word).
X = vectorizer.fit_transform(corpus)

# ---------------------------------------------------------
# 4. PRINT THE TF-IDF FEATURE MATRIX
# ---------------------------------------------------------
print("TF-IDF Feature Matrix:")

# X is stored in a special "sparse" format to save memory.
# toarray() converts it to a regular 2D NumPy array so we can see the numbers.
print(X.toarray())

# (Optional) If you want to see which word each column represents, you can print:
print(vectorizer.get_feature_names_out())
```

This code builds a simple sentiment analysis pipeline that includes feature extraction and model training. It first converts text reviews into TF-IDF numeric features, then splits the data into training and testing sets. A Logistic Regression model is trained on the training data to learn how to classify reviews as positive or negative, and finally the code evaluates the trained model’s performance using accuracy and a detailed classification report.

```python
# ---------------------------------------------------------
# IMPORT THE LIBRARIES WE NEED
# ---------------------------------------------------------
# TfidfVectorizer: turns text into numeric features (TF-IDF scores)
from sklearn.feature_extraction.text import TfidfVectorizer

# train_test_split: splits our data into training and testing sets
from sklearn.model_selection import train_test_split

# LogisticRegression: a simple machine learning model for classification
from sklearn.linear_model import LogisticRegression

# accuracy_score, classification_report: tools to measure how well our model works
from sklearn.metrics import accuracy_score, classification_report


# ---------------------------------------------------------
# 1. CREATE A SMALL TEXT DATASET (CORPUS) AND LABELS
# ---------------------------------------------------------
# "corpus" is a list of text documents (here, short reviews).
corpus = [
    "I love this product! It's amazing.",
    "This is the worst service I have ever experienced.",
    "I am very happy with my purchase.",
    "I am disappointed with the quality of this item."
]

# "labels" tell us whether each text is positive (1) or negative (0).
# These are our "target" values that the model will learn to predict.
labels = [1, 0, 1, 0]  # 1 for positive, 0 for negative


# ---------------------------------------------------------
# 2. CONVERT TEXT INTO NUMBERS USING TF-IDF
# ---------------------------------------------------------
# Machine learning models cannot work directly with raw text.
# TfidfVectorizer converts each sentence into a vector of numbers.
vectorizer = TfidfVectorizer()

# fit_transform() learns the vocabulary from the corpus (fit)
# and then transforms each text into a TF-IDF feature vector (transform).
X = vectorizer.fit_transform(corpus)
# Now "X" is a matrix of shape (number_of_texts, number_of_words_in_vocabulary).


# ---------------------------------------------------------
# 3. SPLIT THE DATA INTO TRAINING AND TESTING SETS
# ---------------------------------------------------------
# We want to train the model on some data and then test it on unseen data.
# test_size=0.25 means 25% of the data is used for testing, 75% for training.
# random_state=42 just makes the split reproducible (same split every time).
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.25, random_state=42
)


# ---------------------------------------------------------
# 4. CREATE AND TRAIN THE LOGISTIC REGRESSION MODEL
# ---------------------------------------------------------
# Logistic Regression is a common model for classification problems
# (like predicting positive vs negative sentiment).
model = LogisticRegression()

# model.fit() lets the model "learn" from the training data.
# It looks at X_train (the features) and y_train (the correct labels).
model.fit(X_train, y_train)


# ---------------------------------------------------------
# 5. MAKE PREDICTIONS ON THE TEST SET
# ---------------------------------------------------------
# Now we use the trained model to predict the labels of X_test,
# which the model has not seen during training.
y_pred = model.predict(X_test)


# ---------------------------------------------------------
# 6. EVALUATE HOW WELL THE MODEL DID
# ---------------------------------------------------------
# accuracy_score: fraction of correct predictions (between 0 and 1).
accuracy = accuracy_score(y_test, y_pred)

# classification_report: gives more detailed metrics, such as
# precision, recall, and F1-score for each class (0 and 1).
report = classification_report(y_test, y_pred)

# Print out the results
print(f"Accuracy: {accuracy}")
print("Classification Report:")
print(report)
```

This code focuses on Evaluating Machine Learning Models for sentiment analysis. It uses a trained model to make predictions on test data, then calculates four key evaluation metrics — accuracy, precision, recall, and F1 score — to measure how well the model is performing and prints these results for easy interpretation.

```python
# ---------------------------------------------------------
# IMPORT METRICS FOR EVALUATING THE MODEL
# ---------------------------------------------------------
# These functions help us measure how good our model is:
# - accuracy_score : how often the model is correct overall
# - precision_score: when the model says "positive", how often is it right?
# - recall_score   : out of all the real "positive" examples, how many did it find?
# - f1_score       : a single score that balances precision and recall
# ---------------------------------------------------------
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# ---------------------------------------------------------
# 1. USE THE TRAINED MODEL TO MAKE PREDICTIONS
# ---------------------------------------------------------
# We use the model to predict the sentiment labels for X_test,
# which is our test set (data the model did NOT see during training).
# y_pred will be a list/array of predicted labels (0s and 1s).
y_pred = model.predict(X_test)

# ---------------------------------------------------------
# 2. CALCULATE EVALUATION METRICS
# ---------------------------------------------------------
# accuracy_score: proportion of correct predictions
accuracy = accuracy_score(y_test, y_pred)

# precision_score: of all the times the model predicted "positive" (1),
# how many of those were actually positive?
precision = precision_score(y_test, y_pred)

# recall_score: of all the actual positive examples in y_test,
# how many did the model correctly predict as positive?
recall = recall_score(y_test, y_pred)

# f1_score: the harmonic mean of precision and recall.
# It gives a single number that balances both.
f1 = f1_score(y_test, y_pred)

# ---------------------------------------------------------
# 3. PRINT THE RESULTS
# ---------------------------------------------------------
print(f"Accuracy: {accuracy}")
print(f"Precision: {precision}")
print(f"Recall: {recall}")
print(f"F1 Score: {f1}")
```

This code builds a simple sentiment analysis system using Convolutional Neural Networks (CNNs). It first converts short text reviews into numeric sequences using tokenization and padding. Then it defines and trains a CNN model in Keras (with an Embedding layer, Conv1D, GlobalMaxPooling, and Dense layers) to classify reviews as positive or negative. Finally, it evaluates the model’s accuracy on test data and uses the trained CNN to predict the sentiment of a new review.

```python
# ---------------------------------------------------------
# SENTIMENT ANALYSIS WITH A SIMPLE CNN (CONVOLUTIONAL NEURAL NETWORK)
# ---------------------------------------------------------
# In this example we:
# 1) Turn short text reviews into numbers that a neural network can understand.
# 2) Build a CNN model in Keras to classify reviews as positive or negative.
# 3) Train, evaluate, and use the model to predict sentiment of new text.
# ---------------------------------------------------------

# NumPy is used for working with arrays of numbers.
import numpy as np

# TensorFlow is a deep learning library.
import tensorflow as tf

# Sequential lets us build a model layer by layer.
from tensorflow.keras.models import Sequential

# Layers we will use in our neural network:
# - Dense: a fully connected layer
# - Conv1D: a 1D convolution layer (useful for text sequences)
# - GlobalMaxPooling1D: reduces sequence to a single max value per feature
# - Embedding: turns word indexes into dense vectors (word embeddings)
from tensorflow.keras.layers import Dense, Conv1D, GlobalMaxPooling1D, Embedding

# Tools to turn text into sequences of numbers:
# - Tokenizer: builds a word index and converts text to sequences of integers
# - pad_sequences: makes all sequences the same length by padding or truncating
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# train_test_split: splits data into training and testing sets
from sklearn.model_selection import train_test_split


# ---------------------------------------------------------
# 1. SAMPLE TEXT DATA (CORPUS) AND LABELS
# ---------------------------------------------------------
# corpus: list of short text reviews
corpus = [
    "I love this product! It's amazing.",
    "This is the worst service I have ever experienced.",
    "I am very happy with my purchase.",
    "I am disappointed with the quality of this item."
]

# labels: sentiment for each review
# 1 means positive review, 0 means negative review
labels = [1, 0, 1, 0]


# ---------------------------------------------------------
# 2. TOKENIZE AND PAD THE TEXT DATA
# ---------------------------------------------------------
# We cannot feed raw text into a neural network.
# We must turn words into numbers.

# Tokenizer:
# - num_words=5000 means keep only the 5000 most common words.
# - It will create a word index (word -> integer ID).
tokenizer = Tokenizer(num_words=5000)

# Learn the word index from our text corpus.
tokenizer.fit_on_texts(corpus)

# Convert each sentence into a sequence of integers.
# Example: "I love this product" might become [2, 15, 7, 23] (numbers are just examples).
X = tokenizer.texts_to_sequences(corpus)

# pad_sequences:
# - Makes all sequences the same length (maxlen=10 here).
# - Short sequences will be padded with zeros at the beginning.
# - Long sequences will be cut off to length 10.
X = pad_sequences(X, maxlen=10)


# ---------------------------------------------------------
# 3. SPLIT THE DATA INTO TRAINING AND TESTING SETS
# ---------------------------------------------------------
# We want to train our model on some data and test it on data it has not seen.
# test_size=0.25 means 25 percent of the data is for testing, 75 percent for training.
# random_state=42 makes the split reproducible.
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.25, random_state=42
)


# ---------------------------------------------------------
# 4. DEFINE THE CNN MODEL
# ---------------------------------------------------------
# We will build a simple CNN (Convolutional Neural Network) for text:
# Embedding -> Conv1D -> GlobalMaxPooling1D -> Dense -> Dense
model = Sequential()

# Embedding layer:
# - input_dim=5000  : size of the vocabulary (same as num_words in Tokenizer)
# - output_dim=50   : each word will be represented by a 50-dimensional vector
# - input_length=10 : each input sequence has length 10 (from pad_sequences)
model.add(Embedding(input_dim=5000, output_dim=50, input_length=10))

# Conv1D layer:
# - filters=128: number of output filters (features) in the convolution.
# - kernel_size=5: size of the sliding window (looks at 5 words at a time).
# - activation='relu': activation function that introduces non-linearity.
model.add(Conv1D(filters=128, kernel_size=5, activation='relu'))

# GlobalMaxPooling1D:
# - Takes the maximum value of each feature across the sequence.
# - This turns a sequence of features into a single vector.
model.add(GlobalMaxPooling1D())

# Dense hidden layer:
# - 10 units with ReLU activation for extra learning capacity.
model.add(Dense(10, activation='relu'))

# Output layer:
# - 1 unit because this is binary classification (positive or negative).
# - 'sigmoid' squashes the output to a value between 0 and 1.
#   Values close to 1 mean positive, close to 0 mean negative.
model.add(Dense(1, activation='sigmoid'))


# ---------------------------------------------------------
# 5. COMPILE THE MODEL
# ---------------------------------------------------------
# We choose:
# - optimizer='adam': a popular optimizer that adjusts the learning rate.
# - loss='binary_crossentropy': standard loss for binary classification.
# - metrics=['accuracy']: we want to track accuracy during training.
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])


# ---------------------------------------------------------
# 6. TRAIN THE MODEL
# ---------------------------------------------------------
# model.fit trains the model:
# - X_train: training input data
# - y_train: training labels
# - epochs=5: number of passes through the training data
# - verbose=1: show progress output while training
# - validation_data: data to check performance after each epoch
#
# We convert y_train and y_test to NumPy arrays to match the expected type.
model.fit(
    X_train,
    np.array(y_train),
    epochs=5,
    verbose=1,
    validation_data=(X_test, np.array(y_test))
)


# ---------------------------------------------------------
# 7. EVALUATE THE MODEL ON TEST DATA
# ---------------------------------------------------------
# model.evaluate returns the loss and metrics for the test set.
loss, accuracy = model.evaluate(X_test, np.array(y_test))
print(f"Accuracy: {accuracy}")


# ---------------------------------------------------------
# 8. USE THE MODEL TO PREDICT SENTIMENT OF NEW TEXT
# ---------------------------------------------------------
# Now we try the model on new, unseen text.

# New example sentence
new_text = ["The product is excellent and I love it."]

# Convert the new text into a sequence of integers using the same tokenizer.
new_text_seq = tokenizer.texts_to_sequences(new_text)

# Pad the sequence so it has the same length as the training data (maxlen=10).
new_text_padded = pad_sequences(new_text_seq, maxlen=10)

# Get the prediction from the model.
# The result will be a number between 0 and 1.
prediction = model.predict(new_text_padded)

# Decide if the prediction is positive or negative.
# We use 0.5 as a cutoff:
# - If prediction > 0.5, we say "Positive".
# - Otherwise, we say "Negative".
print("Prediction:", "Positive" if prediction[0][0] > 0.5 else "Negative")
```

This code demonstrates sentiment analysis using Recurrent Neural Networks (RNNs) and Long Short-Term Memory Networks (LSTMs). It first converts text reviews into padded numeric sequences using tokenization. Then it builds and trains an LSTM-based neural network in Keras to learn patterns in word order and classify reviews as positive or negative. Finally, the trained LSTM model is evaluated on test data and used to predict the sentiment of a new sentence.


```python
# ---------------------------------------------------------
# SENTIMENT ANALYSIS WITH AN LSTM (LONG SHORT-TERM MEMORY NETWORK)
# ---------------------------------------------------------
# In this example, we:
# 1) Turn short text reviews into numbers that a neural network can understand.
# 2) Build an LSTM model in Keras to classify reviews as positive or negative.
# 3) Train, evaluate, and use the model to predict the sentiment of new text.
# This is a simple demo with a very small dataset, just to show the steps.
# ---------------------------------------------------------

# NumPy is used for working with arrays of numbers.
import numpy as np

# TensorFlow is a deep learning library.
import tensorflow as tf

# Sequential lets us build a neural network model layer by layer.
from tensorflow.keras.models import Sequential

# Layers we will use:
# - Dense: a fully connected layer
# - LSTM: a type of recurrent neural network layer that remembers information over time
# - Embedding: turns word indexes into dense vectors (word embeddings)
from tensorflow.keras.layers import Dense, LSTM, Embedding

# Tools to turn text into sequences of numbers:
# - Tokenizer: builds a word index and converts text to sequences of integers
# - pad_sequences: makes all sequences the same length by padding or cutting
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# train_test_split: splits data into training and testing sets
from sklearn.model_selection import train_test_split


# ---------------------------------------------------------
# 1. SAMPLE TEXT DATA (CORPUS) AND LABELS
# ---------------------------------------------------------
# corpus is a list of short text reviews.
corpus = [
    "I love this product! It's amazing.",
    "This is the worst service I have ever experienced.",
    "I am very happy with my purchase.",
    "I am disappointed with the quality of this item."
]

# labels store the sentiment for each review.
# 1 means positive review.
# 0 means negative review.
labels = [1, 0, 1, 0]


# ---------------------------------------------------------
# 2. TOKENIZE AND PAD THE TEXT DATA
# ---------------------------------------------------------
# We cannot feed raw text into an LSTM directly.
# We must first convert words into integer IDs.

# Create a Tokenizer.
# num_words=5000 means we will keep only the 5000 most common words.
tokenizer = Tokenizer(num_words=5000)

# Learn the word index from our text corpus.
tokenizer.fit_on_texts(corpus)

# Convert each sentence into a sequence of integers.
# Example: "I love this product" might become [2, 15, 7, 23]
# (these numbers are just examples based on the learned word index).
X = tokenizer.texts_to_sequences(corpus)

# Make all sequences the same length.
# maxlen=10 means each sequence will have length 10.
# Short sequences will be padded with zeros at the beginning.
# Long sequences will be cut so they have only 10 elements.
X = pad_sequences(X, maxlen=10)


# ---------------------------------------------------------
# 3. SPLIT THE DATA INTO TRAINING AND TESTING SETS
# ---------------------------------------------------------
# Training set is used to teach the model.
# Testing set is used to see how well the model works on unseen data.
# test_size=0.25 means 25 percent of the data is used for testing.
# random_state=42 makes the split the same each time you run the code.
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.25, random_state=42
)


# ---------------------------------------------------------
# 4. DEFINE THE LSTM MODEL
# ---------------------------------------------------------
# We are using a Sequential model where we add layers one by one.
model = Sequential()

# Embedding layer:
# - input_dim=5000  is the size of the vocabulary (same as num_words in Tokenizer).
# - output_dim=50   means each word index will be turned into a 50-dimensional vector.
# - input_length=10 means each input sequence has length 10 (from pad_sequences).
model.add(Embedding(input_dim=5000, output_dim=50, input_length=10))

# LSTM layer:
# - 100 units means the LSTM has 100 memory cells.
# LSTMs are good at learning patterns in sequences, such as word order in sentences.
model.add(LSTM(100))

# Dense output layer:
# - 1 unit because this is binary classification (positive vs negative).
# - activation='sigmoid' squashes the output to a value between 0 and 1.
#   Values close to 1 mean positive, and close to 0 mean negative.
model.add(Dense(1, activation='sigmoid'))


# ---------------------------------------------------------
# 5. COMPILE THE MODEL
# ---------------------------------------------------------
# Before training, we must compile the model.
# - optimizer='adam' is a popular choice that adjusts learning rate automatically.
# - loss='binary_crossentropy' is the standard loss function for binary classification.
# - metrics=['accuracy'] means we will track accuracy during training.
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)


# ---------------------------------------------------------
# 6. TRAIN THE MODEL
# ---------------------------------------------------------
# model.fit trains the model on the training data.
# - X_train: training input sequences.
# - y_train: correct labels for those sequences.
# - epochs=5 means the model will see the whole training set 5 times.
# - verbose=1 means it will print progress information while training.
# - validation_data=(X_test, y_test) lets us see how the model performs on test data
#   at the end of each epoch.
#
# We convert y_train and y_test to NumPy arrays so that Keras receives the data
# in the correct format.
model.fit(
    X_train,
    np.array(y_train),
    epochs=5,
    verbose=1,
    validation_data=(X_test, np.array(y_test))
)


# ---------------------------------------------------------
# 7. EVALUATE THE MODEL ON THE TEST DATA
# ---------------------------------------------------------
# model.evaluate returns the loss and accuracy on the test set.
loss, accuracy = model.evaluate(X_test, np.array(y_test))
print(f"Accuracy: {accuracy}")


# ---------------------------------------------------------
# 8. USE THE MODEL TO PREDICT SENTIMENT OF NEW TEXT
# ---------------------------------------------------------
# Now we try the model on a new, unseen sentence.

# New example review.
new_text = ["The product is excellent and I love it."]

# Convert the new text into a sequence of integers using the same tokenizer.
new_text_seq = tokenizer.texts_to_sequences(new_text)

# Pad the sequence so it has length 10, just like the training data.
new_text_padded = pad_sequences(new_text_seq, maxlen=10)

# Get the prediction from the model.
# The model outputs a number between 0 and 1.
prediction = model.predict(new_text_padded)

# Decide if the prediction is positive or negative using a threshold of 0.5.
# If prediction > 0.5, we call it "Positive".
# Otherwise, we call it "Negative".
print("Prediction:", "Positive" if prediction[0][0] > 0.5 else "Negative")
```

This code demonstrates a simple sentiment analysis system using Transformer-Based Models through a BERT tokenizer combined with a lightweight neural network in Keras. It converts text into BERT-style token IDs, splits the data into training and testing sets, and trains a small neural network to classify reviews as positive or negative. The model is then evaluated for accuracy and used to predict the sentiment of new, unseen text.

```python
# ---------------------------------------------------------
# SIMPLE SENTIMENT CLASSIFIER USING BERT TOKENIZER + KERAS
# ---------------------------------------------------------
# In this example we:
# 1) Use a BERT tokenizer to turn text into integer IDs.
# 2) Build a simple neural network in Keras (NOT a full BERT model).
# 3) Train it to classify short reviews as positive or negative.
# ---------------------------------------------------------

# NumPy is used for working with arrays of numbers.
import numpy as np

# TensorFlow / Keras is used to build and train the neural network.
import tensorflow as tf

# BertTokenizer will turn our text into token IDs similar to what BERT expects.
from transformers import BertTokenizer

# train_test_split will split our data into training and testing sets.
from sklearn.model_selection import train_test_split


# ---------------------------------------------------------
# 1. SAMPLE TEXT DATA (CORPUS) AND LABELS
# ---------------------------------------------------------
# "corpus" is a list of short review sentences.
corpus = [
    "I love this product! It's amazing.",
    "This is the worst service I have ever experienced.",
    "I am very happy with my purchase.",
    "I am disappointed with the quality of this item."
]

# "labels" store the sentiment for each review:
# 1 means positive, 0 means negative.
labels = [1, 0, 1, 0]


# ---------------------------------------------------------
# 2. INITIALIZE THE BERT TOKENIZER
# ---------------------------------------------------------
# We use a pre-trained BERT tokenizer called "bert-base-uncased".
# "uncased" means it does not care about uppercase vs lowercase letters.
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')


# ---------------------------------------------------------
# 3. TOKENIZE AND ENCODE THE TEXT DATA
# ---------------------------------------------------------
# We must convert text into numbers (token IDs) before feeding it to a model.

# max_length is the maximum number of tokens per sentence.
# Sentences longer than this will be cut off, and shorter ones will be padded.
max_length = 10

# tokenizer(...) will:
#  - split each sentence into tokens,
#  - convert tokens to integer IDs,
#  - pad or truncate them to length max_length,
#  - return NumPy arrays because we set return_tensors='np'.
X = tokenizer(
    corpus,
    padding='max_length',      # pad all sequences to the same length
    truncation=True,           # cut off longer sequences
    max_length=max_length,     # fixed length for each sequence
    return_tensors='np'        # return NumPy arrays
)

# input_ids are the actual token IDs that represent our text.
input_ids = X['input_ids']


# ---------------------------------------------------------
# 4. SPLIT DATA INTO TRAINING AND TESTING SETS
# ---------------------------------------------------------
# We split input_ids and labels into:
#  - X_train, y_train: used to train the model
#  - X_test,  y_test : used to test how well the model generalizes
X_train, X_test, y_train, y_test = train_test_split(
    input_ids,
    labels,
    test_size=0.25,       # 25% test, 75% train
    random_state=42       # fixed seed so we get the same split each time
)


# ---------------------------------------------------------
# 5. BUILD A SIMPLE KERAS MODEL
# ---------------------------------------------------------
# NOTE: This is NOT a full BERT model.
# We are only using the BERT tokenizer + a small neural network.

# The vocabulary size is how many different token IDs the tokenizer can use.
vocab_size = tokenizer.vocab_size

# We build a simple model:
#  1) Embedding layer: turns token IDs into dense vectors.
#  2) GlobalAveragePooling1D: averages over all token positions.
#  3) Dense hidden layer: learns some patterns.
#  4) Dense output layer: outputs 2 values (for class 0 and class 1).
model = tf.keras.Sequential([
    # Embedding layer:
    # - input_dim=vocab_size: how many different tokens there are.
    # - output_dim=32: each token will be represented by a 32-length vector.
    # - input_length=max_length: each input sequence has length 10.
    tf.keras.layers.Embedding(input_dim=vocab_size, output_dim=32, input_length=max_length),

    # GlobalAveragePooling1D:
    # - This takes the average of the embeddings across the sequence length.
    # - It compresses the (sequence_length, 32) embeddings into a single 32-length vector.
    tf.keras.layers.GlobalAveragePooling1D(),

    # Dense hidden layer with 16 units and ReLU activation.
    # - This allows the model to learn non-linear patterns.
    tf.keras.layers.Dense(16, activation='relu'),

    # Output layer:
    # - 2 units for 2 classes: class 0 (negative) and class 1 (positive).
    # - softmax activation returns a probability distribution over the 2 classes.
    tf.keras.layers.Dense(2, activation='softmax')
])


# ---------------------------------------------------------
# 6. COMPILE THE MODEL
# ---------------------------------------------------------
# We must compile the model before training it.
# - optimizer='adam': adjusts learning rate automatically and works well in practice.
# - loss='sparse_categorical_crossentropy':
#     used for multi-class (here 2-class) classification when labels are integers (0, 1, 2, ...).
# - metrics=['accuracy']: we want to track how accurate the model is.
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)


# ---------------------------------------------------------
# 7. TRAIN THE MODEL
# ---------------------------------------------------------
# model.fit trains the model on the training data.
# - X_train: training inputs (token IDs)
# - y_train: training labels (0 or 1)
# - epochs=10: the model will go through the entire training data 10 times
# - batch_size=2: number of samples per gradient update
# - validation_data=(X_test, y_test): we also check performance on test data each epoch
model.fit(
    X_train,
    np.array(y_train),
    epochs=10,
    batch_size=2,
    validation_data=(X_test, np.array(y_test))
)


# ---------------------------------------------------------
# 8. EVALUATE THE MODEL
# ---------------------------------------------------------
# model.evaluate measures the loss and accuracy on the test set.
loss, accuracy = model.evaluate(X_test, np.array(y_test))
print(f"Accuracy: {accuracy}")


# ---------------------------------------------------------
# 9. PREDICT THE SENTIMENT OF NEW TEXT
# ---------------------------------------------------------
# Now we try the trained model on a new sentence.

# New example sentence.
new_text = ["The product is excellent and I love it."]

# Tokenize and encode the new text using the same tokenizer.
new_text_enc = tokenizer(
    new_text,
    padding='max_length',
    truncation=True,
    max_length=max_length,
    return_tensors='np'
)

# Make predictions for the new input IDs.
prediction = model.predict(new_text_enc['input_ids'])

# prediction is a 2-element array: [probability_of_class_0, probability_of_class_1]
predicted_class = np.argmax(prediction)  # index of the highest probability (0 or 1)

# Convert the predicted class into a human-readable label.
print("Prediction:", "Positive" if predicted_class == 1 else "Negative")
```
