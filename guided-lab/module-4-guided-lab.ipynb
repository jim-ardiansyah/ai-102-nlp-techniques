This code shows how to turn text into numbers using the Bag of Words method with CountVectorizer. It takes two sentences, learns the unique words (vocabulary), and creates a matrix where each row represents a sentence and each column represents a word. The values in the matrix are word counts, showing how many times each word appears in each sentence.


```python
# Import CountVectorizer from scikit-learn.
# CountVectorizer turns text into numbers by counting how many times each word appears.
from sklearn.feature_extraction.text import CountVectorizer

# Sample text data (also called a "corpus").
# Here we have a list of two short sentences (documents).
documents = [
    "Natural language processing is fun",
    "Language models are important in NLP"
]

# Create (initialize) a CountVectorizer object.
# This object will learn the vocabulary (unique words) and count word frequencies.
vectorizer = CountVectorizer()

# Fit the vectorizer on the text data and transform the documents into a matrix.
# - fit() learns the vocabulary from the documents
# - transform() converts the documents into a numeric matrix
# fit_transform() does both steps at once.
X = vectorizer.fit_transform(documents)

# Convert the matrix into a regular NumPy array so it is easier to see when printed.
bow_array = X.toarray()

# Get the list of feature names (the vocabulary).
# These are all the unique words found in the documents, sorted in order.
vocab = vectorizer.get_feature_names_out()

# Print the vocabulary so we can see which column corresponds to which word.
print("Vocabulary:")
print(vocab)

# Print a blank line and then the Bag of Words array.
# Each row represents one document.
# Each column represents one word from the vocabulary.
# The numbers show how many times each word appears in each document.
print("\nBag of Words Array:")
print(bow_array)
```

This code builds a simple text classifier using word counts and Naive Bayes. It first converts sentences into a Bag of Words matrix with CountVectorizer, then splits the data into training and test sets. A MultinomialNB model is trained on the training data to learn the difference between NLP-related and general AI sentences, and finally the code predicts labels on the test set and prints the model’s accuracy.

```python
# Import CountVectorizer to turn text into numbers (word counts)
from sklearn.feature_extraction.text import CountVectorizer

# Import a Naive Bayes classifier that works well with word counts
from sklearn.naive_bayes import MultinomialNB

# Import a helper function to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import a function to measure how accurate our predictions are
from sklearn.metrics import accuracy_score

# ---------------------------
# 1. SAMPLE DATA
# ---------------------------

# A small list of example text documents (our dataset)
documents = [
    "Natural language processing is fun",
    "Language models are important in NLP",
    "I enjoy learning about artificial intelligence",
    "Machine learning and NLP are closely related",
    "Deep learning is a subset of machine learning"
]

# Labels for each document:
# 1 = NLP-related, 0 = AI-related (but not specifically NLP)
labels = [1, 1, 0, 1, 0]

# ---------------------------
# 2. TEXT TO NUMBERS (VECTORIZATION)
# ---------------------------

# Create (initialize) a CountVectorizer object.
# This will build a vocabulary and count how often each word appears.
vectorizer = CountVectorizer()

# Learn the vocabulary from the documents and transform them into a matrix of word counts.
# X will be a matrix where:
#   - each row is a document
#   - each column is a word from the vocabulary
#   - each value is how many times that word appears in that document
X = vectorizer.fit_transform(documents)

# ---------------------------
# 3. TRAIN / TEST SPLIT
# ---------------------------

# Split the data into training and testing sets.
# test_size=0.2 means 20% of the data will be used for testing, 80% for training.
# random_state=42 makes the split reproducible (same split every time).
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.2, random_state=42
)

# ---------------------------
# 4. CHOOSE AND TRAIN A MODEL
# ---------------------------

# Create (initialize) the Multinomial Naive Bayes classifier.
# This model works well with word count features.
classifier = MultinomialNB()

# Train (fit) the classifier using the training data.
# The model learns the relationship between word counts (X_train)
# and the labels (y_train).
classifier.fit(X_train, y_train)

# ---------------------------
# 5. MAKE PREDICTIONS
# ---------------------------

# Use the trained model to predict labels for the test data.
y_pred = classifier.predict(X_test)

# ---------------------------
# 6. EVALUATE THE MODEL
# ---------------------------

# Calculate how many of the predictions were correct.
# accuracy_score compares the true labels (y_test) with our predictions (y_pred).
accuracy = accuracy_score(y_test, y_pred)

# Print the accuracy so we can see how well the model performed.
print("Accuracy:", accuracy)
```

This code uses TfidfVectorizer to convert a small set of sentences into TF-IDF features, which measure how important each word is in each document. It learns the vocabulary from the text, creates a TF-IDF matrix, then prints the list of words (vocabulary) and the TF-IDF array showing each document as a row and each word’s importance as a column value.

```python
# Import TfidfVectorizer from scikit-learn.
# TF-IDF stands for "Term Frequency – Inverse Document Frequency".
# It turns text into numbers based on how important each word is in the whole set of documents.
from sklearn.feature_extraction.text import TfidfVectorizer

# Sample text data (a "corpus").
# Each string in the list is one document.
documents = [
    "Natural language processing is fun",
    "Language models are important in NLP",
    "I enjoy learning about artificial intelligence",
    "Machine learning and NLP are closely related",
    "Deep learning is a subset of machine learning"
]

# Create (initialize) a TfidfVectorizer object.
# This will learn the vocabulary and calculate TF-IDF scores for each word.
vectorizer = TfidfVectorizer()

# Fit the vectorizer on the text data and transform the documents.
# - fit() learns the vocabulary from the documents
# - transform() calculates TF-IDF values
# fit_transform() does both steps in one line.
X = vectorizer.fit_transform(documents)

# Convert the TF-IDF matrix into a regular NumPy array so it is easier to see when printed.
tfidf_array = X.toarray()

# Get the list of feature names (the vocabulary).
# These are all the unique words found in the documents, sorted in order.
vocab = vectorizer.get_feature_names_out()

# Print the vocabulary so we know which column in the array belongs to which word.
print("Vocabulary:")
print(vocab)

# Print a blank line and then the TF-IDF array.
# Each row represents one document.
# Each column represents one word from the vocabulary.
# Each number shows how important that word is in that document (higher = more important).
print("\nTF-IDF Array:")
print(tfidf_array)
```

This code builds a simple text classification model using TF-IDF features and a Naive Bayes classifier. It converts each sentence into TF-IDF values that measure how important each word is, splits the data into training and test sets, trains the Naive Bayes model on the training data, predicts labels for the test data, and then prints the model’s accuracy.

```python
# Import TfidfVectorizer to turn text into numeric features using TF-IDF
# TF-IDF tells us how important a word is in a document compared to all documents
from sklearn.feature_extraction.text import TfidfVectorizer

# Import a Naive Bayes classifier that works well with text data
from sklearn.naive_bayes import MultinomialNB

# Import a helper function to split data into training and testing sets
from sklearn.model_selection import train_test_split

# Import a function to measure how accurate our predictions are
from sklearn.metrics import accuracy_score

# ---------------------------
# 1. SAMPLE DATA
# ---------------------------

# A small list of example text documents (our dataset)
documents = [
    "Natural language processing is fun",
    "Language models are important in NLP",
    "I enjoy learning about artificial intelligence",
    "Machine learning and NLP are closely related",
    "Deep learning is a subset of machine learning"
]

# Labels for each document:
# 1 means the sentence is NLP related
# 0 means the sentence is AI related but not specifically NLP focused
labels = [1, 1, 0, 1, 0]

# ---------------------------
# 2. TEXT TO NUMBERS WITH TF-IDF
# ---------------------------

# Create (initialize) a TfidfVectorizer object
# This will:
#   1. Build a vocabulary of all words
#   2. Calculate TF-IDF scores for each word in each document
vectorizer = TfidfVectorizer()

# Learn the vocabulary from the documents and transform them into a TF-IDF matrix
# X will be a matrix where:
#   each row is a document
#   each column is a word
#   each value is the TF-IDF score for that word in that document
X = vectorizer.fit_transform(documents)

# ---------------------------
# 3. SPLIT INTO TRAINING AND TESTING SETS
# ---------------------------

# Split the data into training and testing sets
# test_size=0.2 means 20 percent of the data is used for testing, 80 percent for training
# random_state=42 makes the split the same every time you run the code
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.2, random_state=42
)

# ---------------------------
# 4. CHOOSE AND TRAIN A MODEL
# ---------------------------

# Create (initialize) the Multinomial Naive Bayes classifier
classifier = MultinomialNB()

# Train (fit) the classifier using the training data
# The model learns the relationship between TF-IDF features (X_train)
# and the correct labels (y_train)
classifier.fit(X_train, y_train)

# ---------------------------
# 5. MAKE PREDICTIONS
# ---------------------------

# Use the trained model to predict labels for the test data
y_pred = classifier.predict(X_test)

# ---------------------------
# 6. EVALUATE THE MODEL
# ---------------------------

# Calculate the accuracy of the model
# accuracy_score compares the true labels (y_test) with our predictions (y_pred)
accuracy = accuracy_score(y_test, y_pred)

# Print the accuracy so we can see how well the model performed
print("Accuracy:", accuracy)
```

This code trains a simple Word2Vec model on a short text about NLP and AI. It first splits the text into sentences and then into words, trains Word2Vec to learn a 100-dimensional vector for each word, then prints the vector for the word "language" and a list of words that the model thinks are most similar to "language".

```python
# Import the Word2Vec model from the gensim library.
# Word2Vec learns vector (number) representations of words based on how they appear in sentences.
from gensim.models import Word2Vec

# Import sentence and word tokenizers from NLTK.
# sent_tokenize() will split text into sentences.
# word_tokenize() will split sentences into individual words.
from nltk.tokenize import sent_tokenize, word_tokenize

# Import NLTK so we can download needed resources.
import nltk

# Download the "punkt" tokenizer data.
# This is required for sent_tokenize() and word_tokenize() to work.
nltk.download('punkt')

# Sometimes "punkt_tab" is also needed in certain environments.
nltk.download('punkt_tab')

# ---------------------------------------
# 1. SAMPLE TEXT CORPUS
# ---------------------------------------

# This is our sample text. It contains several sentences about NLP and AI.
text = (
    "Natural language processing is fun and exciting. "
    "Language models are important in NLP. "
    "I enjoy learning about artificial intelligence. "
    "Machine learning and NLP are closely related. "
    "Deep learning is a subset of machine learning."
)

# ---------------------------------------
# 2. TOKENIZE INTO SENTENCES
# ---------------------------------------

# Split the long text into a list of sentences.
# Each item in "sentences" is a complete sentence string.
sentences = sent_tokenize(text)

# ---------------------------------------
# 3. TOKENIZE EACH SENTENCE INTO WORDS
# ---------------------------------------

# For each sentence, split it into a list of words.
# word_tokenize() handles spaces and punctuation.
# The result "tokenized_sentences" is a list of lists:
#   [
#     ["Natural", "language", "processing", ...],
#     ["Language", "models", "are", ...],
#     ...
#   ]
tokenized_sentences = [word_tokenize(sentence) for sentence in sentences]

# ---------------------------------------
# 4. TRAIN THE WORD2VEC MODEL
# ---------------------------------------

# Create and train a Word2Vec model.
# - sentences: the tokenized sentences we prepared
# - vector_size: length of the word vector (how many numbers represent each word)
# - window: how many words to the left and right to look at for context
# - sg=1: use the Skip-Gram method (sg=0 would use CBOW)
# - min_count=1: include all words that appear at least 1 time
model = Word2Vec(
    sentences=tokenized_sentences,
    vector_size=100,
    window=5,
    sg=1,
    min_count=1
)

# ---------------------------------------
# 5. GET VECTOR FOR A SPECIFIC WORD
# ---------------------------------------

# Get the vector representation (embedding) for the word "language".
# This returns a NumPy array with 100 numbers (because vector_size=100).
vector = model.wv['language']

print("Vector representation of 'language':")
print(vector)

# ---------------------------------------
# 6. FIND MOST SIMILAR WORDS
# ---------------------------------------

# Find the words that are most similar to "language"
# according to the Word2Vec model.
# The result is a list of (word, similarity_score) pairs.
similar_words = model.wv.most_similar('language')

print("\nMost similar words to 'language':")
print(similar_words)
```

This code loads a pre-trained GloVe word embedding model and uses it to explore word meaning. It retrieves the 100-dimensional vector for the word "language" and then prints a list of words that the model considers most similar to "language" based on their embeddings.

```python
# Import the gensim "downloader" module.
# This lets us easily download and load pre-trained word embedding models.
import gensim.downloader as api

# ---------------------------------------
# 1. LOAD A PRE-TRAINED WORD EMBEDDING MODEL
# ---------------------------------------

# Load the pre-trained GloVe model:
# "glove-wiki-gigaword-100" means:
#   - GloVe embeddings trained on Wikipedia and Gigaword text
#   - each word is represented by a vector of length 100
#
# Note: The first time you run this, it will download the model from the internet.
glove_model = api.load("glove-wiki-gigaword-100")

# ---------------------------------------
# 2. GET THE VECTOR FOR A SINGLE WORD
# ---------------------------------------

# Get the vector representation of the word "language".
# This gives us a NumPy array with 100 numbers that represent the meaning of the word.
vector = glove_model['language']

print("Vector representation of 'language':")
print(vector)

# ---------------------------------------
# 3. FIND WORDS THAT ARE SIMILAR IN MEANING
# ---------------------------------------

# Find the words that are most similar to "language"
# according to the GloVe model.
# The result is a list of (word, similarity_score) pairs.
similar_words = glove_model.most_similar('language')

print("\nMost similar words to 'language':")
print(similar_words)
```

This code loads a pre-trained BERT model and uses it to create numerical embeddings for sentences. It tokenizes each sentence, passes it through BERT, and extracts the embedding of the special [CLS] token, which represents the meaning of the whole sentence. The program then prints part of each embedding and its size.

```python
# ---------------------------------------
# 1. IMPORT LIBRARIES
# ---------------------------------------

# Import the BERT tokenizer and BERT model from the transformers library.
# The tokenizer: turns text (words) into numbers that the model can understand.
# The model: takes those numbers and returns "embeddings" (vectors that represent meaning).
from transformers import BertTokenizer, BertModel

# Import PyTorch, the library used to run the BERT model.
import torch

# ---------------------------------------
# 2. LOAD PRE-TRAINED BERT MODEL & TOKENIZER
# ---------------------------------------

# Load a pre-trained BERT tokenizer.
# "bert-base-uncased" is a common version of BERT that:
# - works with English
# - converts everything to lowercase (uncased)
tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")

# Load the pre-trained BERT model that matches the tokenizer above.
# This model has already been trained on a large English text dataset.
model = BertModel.from_pretrained("bert-base-uncased")

# Put the model into "evaluation" mode.
# This tells the model we are only using it to get outputs, not to train it.
model.eval()

# ---------------------------------------
# 3. SAMPLE TEXT DATA
# ---------------------------------------

# A small list of sentences we want to turn into BERT embeddings.
sentences = [
    "Natural Language Processing is fascinating.",
    "BERT embeddings help computers understand text better.",
    "I am learning how to use BERT in Python."
]

# ---------------------------------------
# 4. FUNCTION TO GET BERT EMBEDDINGS
# ---------------------------------------

def get_bert_embedding(sentence):
    """
    Take one sentence (a string) and return its BERT embedding.
    We use the [CLS] token embedding, which is often used to represent
    the whole sentence in BERT.
    """

    # Use the tokenizer to convert the sentence into model inputs.
    # This step:
    # - splits the sentence into tokens (sub-words)
    # - converts tokens into numeric IDs
    # - creates an attention mask so BERT knows which tokens are real and which are padding
    # return_tensors="pt" means "return the result as PyTorch tensors".
    inputs = tokenizer(
        sentence,
        return_tensors="pt",   # Return PyTorch tensors (required by the model)
        truncation=True,       # Safely cut off very long sentences if needed
        padding=True           # Add padding if needed (useful for batches)
    )

    # We only want to run the model to get outputs, not to train it.
    # torch.no_grad() turns off gradient tracking to make it faster and use less memory.
    with torch.no_grad():
        # Run the model on the tokenized inputs.
        # This returns an object that contains several things,
        # including "last_hidden_state".
        outputs = model(**inputs)

    # last_hidden_state is a 3D tensor with shape:
    # (batch_size, sequence_length, hidden_size)
    # - batch_size: how many sentences (here it's 1)
    # - sequence_length: how many tokens in the sentence
    # - hidden_size: size of each embedding vector (usually 768 for BERT base)
    #
    # The [CLS] token is the first token in every input sentence,
    # so we take position 0 along the sequence dimension.
    cls_embedding = outputs.last_hidden_state[:, 0, :]  # shape: (1, hidden_size)

    # Remove the batch dimension (since we only have 1 sentence).
    # This gives us a 1D vector: just the embedding for this sentence.
    return cls_embedding.squeeze(0)  # shape: (hidden_size,)

# ---------------------------------------
# 5. GENERATE AND PRINT EMBEDDINGS
# ---------------------------------------

# Loop over each sentence and show its BERT embedding.
for i, sentence in enumerate(sentences, start=1):
    # Get the [CLS] embedding for this sentence
    embedding = get_bert_embedding(sentence)

    print(f"\nSentence {i}: {sentence}")
    print("BERT [CLS] Embedding (first 10 values):")
    # Print only the first 10 numbers from the embedding vector
    # so the output is not too long or overwhelming.
    print(embedding[:10])

    # Print the length of the embedding vector.
    # For bert-base-uncased, this is usually 768.
    print("Embedding size:", embedding.shape[0])
```
